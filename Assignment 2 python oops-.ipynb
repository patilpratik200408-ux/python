```python
class AnswerSheet:
    bonusmarks =0
    def __init__(self):
        self.id =input('Enter Id:')
        self.marks =int(input('Enter Marks:'))
        self.totalmarks = self.marks
        AnswerSheet.bonusmarks =  AnswerSheet.bonusmarks +5
    def __sub__(self,x):
        self.totalmarks = self.marks - x
    def __add__(self,x):
        self.totalmarks = self.marks + x
    def __len__(self):
        return len(self.id)
    def __del__(self):
        print("AnswerSheet object deleted")
    
    
pavan = AnswerSheet()
pavan -20
print(pavan.totalmarks+pavan.bonusmarks)
print(len(pavan))
del pavan
ram = AnswerSheet()
ram + 11
print(ram.totalmarks+ram.bonusmarks)
```
