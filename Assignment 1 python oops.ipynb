```python
class BankBalance:
    def __init__(self):
        self.accountnumber =input('Enter Account Number:')
        self.Balance =int(input('Enter Total Balance :'))
        self.totalbalance = self.Balance
    def debit(self):
        debt = int(input('Enter How Many Amount You Do To Withdraw:'))
        self.Balance = self.Balance - debt
        self.totalbalance = self.Balance
    def credit(self):
        credit = int(input("Enter How Many Amout Do you Deposit: "))
        self.Balance = self.Balance +credit
        self.totalbalance = self.Balance
    
pavan = BankBalance()
pavan.debit()
print(pavan.totalbalance)

ram = BankBalance()
print(ram.totalbalance)

rohan = BankBalance()
rohan.credit()
print(rohan.totalbalance)

vishal =BankBalance()
print(vishal.totalbalance)

vikrant = BankBalance()
vikrant.debit()
print(vikrant.totalbalance)
```
