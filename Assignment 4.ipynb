```python
class Employee:
    def __init__(self,empid,empname,emprole,empsalary):
        self.empid = empid
        self.empname = empname
        self.emprole = emprole
        self.empsalary = empsalary
    def get_role(self):
        return "role : Employee"
class Developer(Employee):
    def __init__(self,empid,empname,emprole,empsalary):
        super().__init__(empid,empname,emprole,empsalary)
    def get_role(self):
        return "role : Developer - Software Developer"
class Datascientist(Employee):
    def __init__(self,empid,empname,emprole,empsalary):
        super().__init__(empid,empname,emprole,empsalary)
    def get_role(self):
        return "role : Data Scientist - Machine Learning Enginer"
class Manager(Employee):
    def __init__(self,empid,empname,emprole,empsalary):
        super().__init__(empid,empname,emprole,empsalary)
    def get_role(self):
        return "role: Manager : Team Manager"

class Task:
    def __init__(self,taskid,empid,task):
        self.taskid = taskid
        self.empid = empid
        self.task = task
        self.status = "Pending"
    def display(self):
        return f"Task NO:{self.taskid},Employee Id:{self.empid},Task :{self.task},Status:{self.status}"
def taskassign():
    print('task assigned')
def decorate(f):
    def f1():
        print("Welcome Task is ....")
        f()
    return f1
@decorate
def taskassign():
    print("Task assigned")

def taskgen(tasks):
        for task in tasks:
            yield task 
emp =[]
tasks=[]
dev = Developer(1,"rohan","software Developer",400000)
emp.append(dev)
ds = Datascientist(2,'kartik sharma','machine Learning engineer',500000)
emp.append(ds)

task1 = Task(1,2,"build machine learning model")
tasks.append(task1)
task2 = Task(2,3,"build API keys:")
for employee in emp:
    print("ID:", employee.empid)
    print("Name:", employee.empname)
    print("Role:", employee.emprole)
    print("Salary:", employee.empsalary)
tasks.append(task2)
print()
print(task1.display())
print(task2.display())

print()
print("Emp Names:")
itr1 = iter(emp)
print(next(itr1).empname)
print(next(itr1).empname)
print()
taskassign()
print()
gen = taskgen(tasks)
print(next(gen).task)
print(next(gen).task)

```
